```python
import pandas as pd
import matplotlib.dates as mdates
from matplotlib import pyplot as plt
import numpy as np

from datetime import datetime, timedelta
```

```python
sunspots = pd.read_csv('sunspot_data_processed.csv')
sunspots = sunspots.drop(columns=['Monthly mean', 'Number of observations', 'Definitive or provisional value'])

sunspots.head()
```

```python
sunspots.info()
```

```python
sunspots.describe()
```

```python
# Plotting the mean sunspot numbers over time
plt.figure(figsize=(12, 6))
plt.scatter(sunspots['Date in fraction of year'], sunspots['Monthly mean total sunspot number'])
plt.title('Monthly Mean Sunspot Numbers Over Time')
plt.xlabel('Year')
plt.ylabel('Monthly Mean Total Sunspot Number')
plt.grid()
plt.tight_layout()
plt.show()
```
